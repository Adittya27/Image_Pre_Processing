```python
img.resize((400,400)) # Resizing the image
```

```python
img.rotate(90) # Rotating the image
```

```python
#Image Mirror

from PIL import ImageOps
img = ImageOps.mirror(img)

img
```

```python
ImageOps.flip(img) #flipping the image
```

```python
cropped=img.crop((50,50,250,250)) #Croping the image
cropped
```

```python
img.convert("L") #Converting image to grayscale
```

```python
from PIL import ImageFilter # converting the image to blured
blur=img.filter(ImageFilter.BLUR)
blur
```

```python
# shapening the image
Sharp = img.filter(ImageFilter.SHARPEN)

Sharp
```

```python
#brightening the image
from PIL import ImageEnhance

enhancer = ImageEnhance.Brightness(img)

bright = enhancer.enhance(2)

bright
```

```python
#Splittin the RGB chanels
r, g, b = img.split()

red_img = Image.merge('RGB', (r, Image.new('L', r.size, 0),
                              Image.new('L', r.size, 0)))
green_img = Image.merge('RGB', (Image.new('L', g.size, 0), g,
                              Image.new('L', g.size, 0)))
blue_img = Image.merge('RGB', (Image.new('L', b.size, 0),
                              Image.new('L', b.size, 0),b))
```
